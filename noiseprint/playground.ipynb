## I - How to use Noiseprint in evaluation mode (using pre-trained weights) ?

You can simply use the function `getNoiseprint` to get the noiseprint of the image located at `img_path`.


I put below the results using the examples for the original authors.

```python
examples=[]
noiseprints=[]
for img_path in glob.glob('examples/*'):
     example,noise_print=getNoiseprint(img_path)
     examples.append(example)
     noiseprints.append(noise_print)
```

```python
plt.figure(figsize=(100,100))
for i,(img,res) in enumerate(zip(examples,noiseprints)):
    plt.subplot(5,2,2*i+1)
    plt.imshow(img,clim=[0,1],cmap='gray')
    vmin = np.min(res[34:-34,34:-34])
    vmax = np.max(res[34:-34,34:-34])
    plt.subplot(5,2,2*i+2)
    plt.imshow(res.clip(vmin,vmax),clim=[vmin,vmax],cmap='gray');
```

```python
plt.figure(figsize=(20,20))
plt.subplot(1,2,1)
img=examples[4]
res=noiseprints[4]
vmin = np.min(res[34:-34,34:-34])
vmax = np.max(res[34:-34,34:-34])
plt.imshow(img,clim=[0,1],cmap='gray');
plt.subplot(1,2,2)
plt.imshow(res.clip(vmin,vmax),clim=[vmin,vmax],cmap='gray');
```
